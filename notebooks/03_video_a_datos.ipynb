# Del video a los datos

Lo mismo que hace el panel, desde Python: calibrar, procesar el video y leer los resultados con el mismo esquema que StatsBomb o SkillCorner. Este ejemplo usa el clip sintético del repositorio (`demo/clips/tv`), que trae la verdad de cada cuadro para medir el error.

```python
from pathlib import Path
from analista.vision import Calibracion, ConfigPipeline, Pipeline, FuenteArchivo

raiz = Path("..")
cal = Calibracion.cargar(raiz / "demo/clips/tv/calibracion.json")
cfg = ConfigPipeline(modelo_yolo=str(raiz / "modelos/yolo11n.pt"), imgsz=1280, ancho_proceso=1280,
                     equipos={"Auckland FC": "#1f6fd1", "Newcastle Jets": "#f2f2f2"})
pipe = Pipeline(cal, cfg)
fuente = FuenteArchivo(str(raiz / "demo/clips/tv/video.mp4"), 0, 6)   # los primeros 6 segundos
resumen = pipe.correr(fuente, Path("salida_tv"), video_anotado=False, progreso=False)
resumen
```
